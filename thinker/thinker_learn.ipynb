## 1. 主窗口与事件循环

每个 tkinter 程序都由三个核心要素组成：**主窗口（Tk 实例）**、**控件树（Widget 层级）**、**事件循环（mainloop）**。

| 步骤 | 代码 | 作用 |
|---|---|---|
| 创建主窗口 | `root = tk.Tk()` | 初始化 Tk，创建根窗口 |
| 设置标题 | `root.title("标题")` | 窗口标题栏文字 |
| 设置大小 | `root.geometry("400x300")` | 宽 x 高（像素），可加位置 `+100+50` |
| 进入事件循环 | `root.mainloop()` | 阻塞运行，响应用户输入、刷新界面 |

**`mainloop()` 是关键**：没有它，窗口创建后会瞬间消失。它让程序进入事件驱动模式，等待并处理鼠标点击、键盘输入等事件[citation:1]。

**窗口关闭**：点击标题栏的 × 默认会终止程序。如需自定义关闭行为，可绑定 `WM_DELETE_WINDOW` 协议。

````python
# 最小可运行窗口（取消注释运行）
root = tk.Tk()
root.title("我的第一个 tkinter 窗口")
root.geometry("400x300+200+100")   # 宽400 高300，位置(200,100)
root.mainloop()

print("取消注释即可看到窗口")
````

## 2. 基础控件

tkinter 的界面由一个个**控件（Widget）**组成，每个控件是一个 Python 对象。创建子控件时，**第一个参数是父容器**（如 `root` 或 `Frame`）[citation:1]。

| 控件 | 类名 | 用途 |
|---|---|---|
| 标签 | `Label` | 显示文本或图像 |
| 按钮 | `Button` | 可点击，通过 `command` 绑定函数 |
| 单行输入 | `Entry` | 获取用户输入，`.get()` 读取 |
| 多行文本 | `Text` | 富文本编辑或日志显示 |
| 复选框 | `Checkbutton` | 多选，绑定 `IntVar`/`BooleanVar` |
| 单选按钮 | `Radiobutton` | 单选，绑定 `StringVar`/`IntVar` |
| 列表框 | `Listbox` | 显示可选列表 |
| 框架 | `Frame` | 容器，用于分组和布局 |

**通用配置选项**：`text`（文字）、`font`（字体）、`bg`/`fg`（背景/前景色）、`width`/`height`（尺寸）、`command`（按钮回调）、`state`（状态：`normal`/`disabled`）[citation:2]。

**变量类**：`IntVar`、`StringVar`、`BooleanVar` 用于在控件和 Python 代码之间同步值，修改后控件自动更新。

````python
# 基础控件演示（取消注释运行）
root = tk.Tk()
root.title("基础控件")
root.geometry("300x200")

tk.Label(root, text="用户名:").pack(pady=5)
entry = tk.Entry(root)
entry.pack(pady=5)

def show_input():
    print("输入:", entry.get())

tk.Button(root, text="获取输入", command=show_input).pack(pady=5)

var = tk.IntVar()
tk.Checkbutton(root, text="同意条款", variable=var).pack(pady=5)

root.mainloop()

print("控件代码已注释")
````

## 3. 布局管理

控件创建后**不会自动显示**，必须通过几何管理器指定位置。tkinter 提供三种布局方式[citation:8]：

| 管理器 | 方法 | 特点 | 适用场景 |
|---|---|---|---|
| `pack()` | 按添加顺序排列 | 简单，灵活性较差 | 快速原型、简单垂直/水平排列 |
| `grid()` | 表格（行/列） | 灵活，**最推荐** | 表单、复杂布局 |
| `place()` | 绝对坐标 | 最灵活但最难维护 | 精确位置控制 |

**`grid()` 常用参数**：`row`（行）、`column`（列）、`sticky`（对齐，`'n'`/`'s'`/`'e'`/`'w'` 或组合如 `'nsew'`）、`columnspan`/`rowspan`（跨列/跨行）、`padx`/`pady`（外边距）[citation:8]。

**布局建议**：优先用 `grid()` 实现复杂布局，用 `Frame` 作为容器分组控件，混合使用 `pack()` 和 `grid()` 时注意不要在同一父容器内混用两种管理器。

**易错点**：`pack()` 和 `grid()` 不能对**同一个父容器**中的控件混用，否则会卡死或布局混乱。

````python
# grid 布局演示（取消注释运行）
root = tk.Tk()
root.title("登录")

tk.Label(root, text="用户名:").grid(row=0, column=0, sticky='e', padx=5, pady=5)
tk.Entry(root).grid(row=0, column=1, padx=5, pady=5)

tk.Label(root, text="密码:").grid(row=1, column=0, sticky='e', padx=5, pady=5)
tk.Entry(root, show='*').grid(row=1, column=1, padx=5, pady=5)

tk.Button(root, text="登录").grid(row=2, column=0, columnspan=2, pady=10)

root.mainloop()

print("grid 布局代码已注释")
````

## 4. 事件绑定

除了按钮的 `command`，tkinter 允许通过 `bind()` 将任意事件绑定到回调函数。

| 事件 | 触发条件 |
|---|---|
| `<Button-1>` | 鼠标左键点击 |
| `<KeyPress-A>` | 按下 A 键 |
| `<Return>` | 按下回车 |
| `<Configure>` | 窗口尺寸变化 |

**回调函数签名**：绑定的事件回调会收到一个 `event` 参数，包含 `x`、`y`（坐标）、`char`（按键字符）等信息。

**`bind()` 的第三个参数 `"+"`**：默认情况下，对同一事件的多次 `bind` 会覆盖之前的回调；传入 `"+"` 表示**追加**而非覆盖[citation:4]。

**易错点**：如果绑定的函数不需要用到 `event`，可以用 `lambda e: func()` 包装，否则函数必须接受一个参数。

````python
# 事件绑定演示（取消注释运行）
root = tk.Tk()
root.title("事件绑定")

def on_click(event):
    print(f"点击坐标: ({event.x}, {event.y})")

def on_enter(event):
    print("按下了回车键")

canvas = tk.Canvas(root, width=300, height=200, bg='white')
canvas.pack()
canvas.bind("<Button-1>", on_click)

entry = tk.Entry(root)
entry.pack()
entry.bind("<Return>", on_enter)

root.mainloop()

print("事件绑定代码已注释")
````

## 5. ttk 主题控件

`tkinter.ttk` 模块提供了一套**带主题的现代化控件**，外观比经典 tkinter 控件更美观，且在不同平台上更接近原生风格[citation:1][citation:5]。

| 经典控件 | ttk 替代 |
|---|---|
| `tk.Label` | `ttk.Label` |
| `tk.Button` | `ttk.Button` |
| `tk.Entry` | `ttk.Entry` |
| `tk.Frame` | `ttk.Frame` |

**ttk 独有控件**：
- `ttk.Combobox`：下拉选择框，`values` 设置选项列表，`.get()` 获取当前值。
- `ttk.Notebook`：标签页容器，每个标签对应一个子窗口。
- `ttk.Treeview`：树形/表格控件，适合展示层级数据或表格数据[citation:5]。
- `ttk.Spinbox`：带增减按钮的输入框（Python 3.7+）。

**建议**：新项目优先使用 `ttk` 控件，外观和平台一致性更好。

````python
# ttk 控件演示（取消注释运行）
root = tk.Tk()
root.title("ttk 控件")

frame = ttk.Frame(root, padding=10)
frame.grid()

ttk.Label(frame, text="选择一个选项:").grid(row=0, column=0, padx=5)
combo = ttk.Combobox(frame, values=["选项A", "选项B", "选项C"])
combo.grid(row=0, column=1, padx=5)

ttk.Button(frame, text="确定", command=lambda: print(combo.get())).grid(row=1, column=0, columnspan=2, pady=10)

root.mainloop()

print("ttk 代码已注释")
````

## 6. 消息框与文件对话框

tkinter 提供了标准对话框模块，无需自己绘制弹窗。

**消息框**（`tkinter.messagebox`）

| 函数 | 用途 |
|---|---|
| `showinfo(title, message)` | 信息提示 |
| `showwarning(title, message)` | 警告提示 |
| `showerror(title, message)` | 错误提示 |
| `askyesno(title, message)` | 是/否确认，返回 `True`/`False` |

**文件对话框**（`tkinter.filedialog`）

| 函数 | 用途 |
|---|---|
| `askopenfilename()` | 选择打开的文件，返回路径字符串 |
| `asksaveasfilename()` | 选择保存位置，返回路径字符串 |
| `askdirectory()` | 选择文件夹 |

**易错点**：对话框函数会**阻塞程序**直到用户操作，不要在不需要交互的场景（如后台任务）中调用。

````python
# 对话框演示（取消注释运行）
from tkinter import messagebox, filedialog

root = tk.Tk()
root.withdraw()  # 隐藏主窗口，只显示对话框

messagebox.showinfo("提示", "这是一个信息框")
result = messagebox.askyesno("确认", "确定要继续吗？")
print("用户选择:", result)

path = filedialog.askopenfilename(
    title="选择文件",
    filetypes=[("文本文件", "*.txt"), ("所有文件", "*.*")]
)
print("选择路径:", path)

print("对话框代码已注释")
````

## 7. 综合练习

1. 写一个简易计算器界面：两个输入框（数字）、一个下拉框（运算符 + - * /）、一个按钮、一个结果标签。
2. 做一个文件选择器：点击按钮选择文件，在 Label 中显示文件路径。
3. 用 `ttk.Notebook` 创建两个标签页，分别放一个“输入”框和一个“显示”框。

**参考答案（简易计算器）**

````python
# 简易计算器（取消注释运行）
root = tk.Tk()
root.title("计算器")

ttk.Label(root, text="数字1:").grid(row=0, column=0, padx=5, pady=5)
num1 = ttk.Entry(root, width=10)
num1.grid(row=0, column=1, padx=5, pady=5)

ttk.Label(root, text="运算符:").grid(row=1, column=0, padx=5, pady=5)
op = ttk.Combobox(root, values=["+", "-", "*", "/"], width=5, state="readonly")
op.grid(row=1, column=1, padx=5, pady=5)
op.set("+")

ttk.Label(root, text="数字2:").grid(row=2, column=0, padx=5, pady=5)
num2 = ttk.Entry(root, width=10)
num2.grid(row=2, column=1, padx=5, pady=5)

result_label = ttk.Label(root, text="结果: -")
result_label.grid(row=3, column=0, columnspan=2, pady=10)

def calculate():
    try:
        a, b = float(num1.get()), float(num2.get())
        if op.get() == "+": r = a + b
        elif op.get() == "-": r = a - b
        elif op.get() == "*": r = a * b
        else: r = a / b if b != 0 else "除数不能为0"
        result_label.config(text=f"结果: {r}")
    except ValueError:
        result_label.config(text="请输入有效数字")

ttk.Button(root, text="计算", command=calculate).grid(row=4, column=0, columnspan=2, pady=5)

root.mainloop()

print("计算器代码已注释")
````

## 常见坑点总结

- **包名是 `tkinter` 不是 `thinker`**：全小写，Python 3 中导入 `import tkinter`。
- **忘记 `mainloop()` 窗口一闪而过**：没有事件循环，程序创建窗口后立即退出。
- **`pack` 和 `grid` 不能混用**：同一个父容器内混用两种几何管理器会导致布局失控。
- **控件创建后不会自动显示**：必须调用 `pack()`/`grid()`/`place()` 之一。
- **`bind` 回调需要接收 `event` 参数**：不需要时用 `lambda e: func()` 包装。
- **`mainloop()` 是阻塞的**：它后面的代码在窗口关闭前不会执行，耗时操作应放在回调函数或线程中。
- **优先用 `ttk` 控件**：外观更现代，跨平台一致性更好。
- **`command` 接收的是函数对象**：写 `command=my_func` 而不是 `command=my_func()`，加括号会立即执行。
